```python
# Stack implementation with limited size

class Stack:

    def __init__(self,maxsize):
        self.s1 = [] 
        self.maxsize = maxsize

    def display(self):
        if self.IsEmpty():
            print("No element in Stack")
        else:
            for i in range(len(self.s1)-1,-1,-1):
                print(self.s1[i])

    def IsFull(self):
        if len(self.s1) == self.maxsize:
            return True
        else:
            return False

    def IsEmpty(self):
        if len(self.s1) == 0 :
            return True
        else:
            return False

    def push(self,val):
        if self.IsFull():
            print("Stack is Full")
        else:
            self.s1.append(val)
            

    def pop(self):
        if self.IsEmpty():
            print("The stack is empty")
        else:
            self.s1.pop()

    def peek_ele(self):
        if self.IsEmpty():
            print("No Element Present in Stack")
        else:
            print(self.s1[-1])

s1 = Stack(5)
while True:
    print()
    choice = int(input("Enter Your Choice: "))
    match choice:
        case 1 :
            print("Is Stack is Full : ",s1.IsFull())
        case 2 :
            print("Is Stack is Empty : ",s1.IsEmpty())
        case 3 :
            val = int(input("Enter the value : " ))
            s1.push(val)
        case 4 :
            s1.pop()
        case 5 :
            s1.peek_ele()
        case 6 :
            s1.display()
        case _ :
            print("Wrong Choice")

```

```python
# Recurssion
# Question --> Sum of natural Numbers ?
def sumofnums(n):
    if n == 1:
        return 1
    else:
        return n +sumofnums(n-1)
sumofnums(5)
```

```python
# Question --> Factorial of a integer ? 

def factorial(n):
    if n == 5:
        return 5
    else:
        return n * factorial(n+1)

factorial(1)
```

```python
# Reverse the Given string using Recurssion

def reverse(n):
    if len(n) == 1:
        return n
    else:
        return n[-1] + reverse(n[:-1])

print(reverse("Python"))
```

```python
# Given parenthesis are Balanced or not using stack


def is_balanced(s1):
    stack = []
    d1 = {
        ')' : '(',
        ']' : '[',
        '}' : '{'
    }

    for char in s1:
        if char in '([{':
            stack.append(char)
        elif char in ')}]':
            if not stack or stack.pop() != d1[char]:
                return False
    return len(stack) == 0


s1 = "[[[({})]]]"
if is_balanced(s1):
    print("Balanced")
else:
    print("Not Balanced")
```
