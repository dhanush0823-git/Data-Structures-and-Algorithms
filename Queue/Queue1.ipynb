## **Linear Queue**

```python
# Queue With maxsize

class Queue:

    def __init__(self,maxsize):
        self.maxsize = maxsize
        self.q1 = []

    def Isfull(self):
        if len(self.q1) == self.maxsize:
            return True
        else:
            return False

    def Isempty(self):
        if len(self.q1) == 0:
            return True
        else:
            return False

    def display(self):
        if self.Isempty():
            print("Queue is Empty")
        else:
            print(self.q1)

    def enqueue(self,val):
        if self.Isfull():
            print("Queue is Full")
        else:
            self.q1.append(val)

    def dequeue(self):
        if self.Isempty():
            print("Queue is Empty")
        else:
            self.q1.pop(0)

    def peek(self):
        if self.Isempty():
            print("Empty")
        else:
            print(self.q1[0])
```

## **Circular Queue**

```python
class CircularQueue:

    def __init__(self,maxsize):
        self.maxsize = maxsize
        self.q1 = [None] * maxsize
        self.front = -1
        self.rear = -1

    def display(self):
        print(self.q1)
        print(self.front,"Front")
        print(self.rear,"Rear")

    def Isfull(self):
        if self.front == 0 and self.rear +1 == self.maxsize:
            return True
        elif self.rear + 1 == self.front:
            return True
        else:
            return False

    def Isempty(self):
        if self.rear == -1 and self.front == -1:
            return True
        else:
            return False

    def Enqueue(self,val):
        if self.Isfull():
            print("Circular Queue is full")
        else:
            if self.rear + 1 == self.maxsize:
                self.rear = 0
            else:
                self.rear += 1
                if self.front == -1:
                    self.front = 0
            self.q1[self.rear] = val

    def Dequeue(self):
        if self.Isempty():
            print("Queue is Empty")
        else:
            idx = self.front
            if self.front == self.rear:
                self.front = -1
                self.rear = -1
            elif self.front +1 == self.maxsize:
                self.front = 0
            else:
                self.front +=1
            self.q1[idx] = None

    def peek_ele(self):
        print(self.q1[self.front])
cq = CircularQueue(5)
# Use Match Case to Execute
cq.Enqueue(10)
cq.Enqueue(20)
cq.Enqueue(30)
cq.Enqueue(40)
cq.Enqueue(50)
cq.display()
cq.Dequeue()
cq.Dequeue()

cq.display()
cq.Enqueue(60)
cq.display()
cq.Enqueue(70)
cq.display()
cq.Dequeue()

cq.display()
cq.peek_ele()
```
